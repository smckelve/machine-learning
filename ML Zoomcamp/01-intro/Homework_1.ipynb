```python
import numpy as np

asia_cars = car_data[car_data["origin"] == "Asia"]
subset = asia_cars[["vehicle_weight", "model_year"]]
first_7 = subset.head(7)
X = first_7.values
X
```

```python
XTX = X.T @ X
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
w = XTX_inv @ X.T @ y
w
```

```python
w_sum = w.sum()
w_sum
```
